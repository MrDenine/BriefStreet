## Setup

```python
# load environment variables
import os
import dotenv

dotenv.load_dotenv(dotenv_path='.env')

apiKey = os.environ['BINANCE_API_KEY']
secret = os.environ['BINANCE_SECRET_KEY']
```

```python
import ccxt
import pandas as pd
```

```python
exchange = ccxt.binance()
exchange.requiredCredentials
```

```python
auth = {
    "apiKey": apiKey,
    "secret": secret
}
exchange = ccxt.binance(auth)
```

## Balance

```python
exchange.fetch_balance()
```

```python
print("BTC:",exchange.fetch_balance()['BTC'])
print("USDT:",exchange.fetch_balance()['USDT'])
```

```python
usdt_available = exchange.fetch_balance()['USDT']['free']
print("USDT available:", usdt_available)
```

## Load Market

```python
markets = exchange.load_markets()
df = pd.DataFrame(markets)
print(df.columns)
print(df)
```

```python
ticker = exchange.fetch_ticker('BTC/USDT')
print(ticker)
```

```python
print('Bid price:', ticker['bid'], 'Ask price:', ticker['ask'])
```

## Place Orders

```python
sumbol = 'BTC/USDT'
order_type = 'limit'
side = 'buy'

base_price = 0.01

amount_in_usdt = 10

price = base_price * 0.99  # Buy at 1% below base price

amount = amount_in_usdt / price

order = exchange.create_order(sumbol, order_type, side, amount, price)
print(order)
```
